```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.png")          
print(img.shape)   # (yükseklik, genişlik, 3) → 3 kanal varsa renkli
print(img.dtype)   # uint8 → 8 bit, değerler 0–255


plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("RGB sırası")

plt.subplot(1,2,2)
plt.imshow(img)
plt.title("BGR sırası")
plt.show()

# cv2 kütüphanesi renkleri BGR şeklinde "img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)"
# şeklinde düzeltip RGB yapmazsak renkler yanlış çıkar
```
